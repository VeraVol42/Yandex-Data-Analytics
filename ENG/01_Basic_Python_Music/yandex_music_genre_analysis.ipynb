**Task 20**

```python
def number_tracks(day, city):
    track_list = df[df['day'] == day] # select only rows from df where the 'day' column equals the variable day
    track_list = track_list[track_list['city'] == city] # select only rows from track_list where the 'city' column equals the variable city
    track_list_count = track_list['user_id'].count() # call the method to count rows for track_list using the 'user_id' column
    return track_list_count # return the value of track_list_count from the function
```

### Music at the Beginning and End of the Week

According to the second hypothesis, on Monday morning in Moscow one set of genres prevails, while in St. Petersburg different genres dominate. Similarly, on Friday evening, different genres prevail depending on the city.

**Task 23**

```python
# obtaining the moscow_general table from those rows of df where the 'city' column equals 'Moscow'
moscow_general = df[df['city'] == 'Moscow']
```

```python
# obtaining the moscow_general table from those rows of df where the 'city' column equals 'Saint-Petersburg'
spb_general = df[df['city'] == 'Saint-Petersburg']
```

**Task 24**

```python
def genre_weekday(df, day, time1, time2):
    # sequential filtering
    # keep only the rows of df where the day equals day
    genre_df = df[df['day'] == day] 
    # keep only the rows of genre_df where the time is less than time2
    genre_df = genre_df[genre_df['time'] < time2]
    # keep only the rows of genre_df where the time is greater than time1
    genre_df = genre_df[genre_df['time'] > time1]
    # group the filtered DataFrame by the column with genre names,
    # take the 'genre' column and count the number of rows for each genre using the count() method
    genre_df_grouped = genre_df.groupby('genre')['genre'].count()
    # sort the result in descending order (so that the most popular genres appear at the beginning of the Series)
    genre_df_sorted = genre_df_grouped.sort_values(ascending=False)
    # return a Series with the top 10 most popular genres in the specified time interval for the given day
    return genre_df_sorted[:10]
```

**Task 25**

```python
genre_weekday(moscow_general, 'Monday', '07:00', '11:00') # call the function for Monday morning in Moscow 
```

```python
genre_weekday(spb_general, 'Monday', '07:00', '11:00') # call the function for Monday morning in St. Petersburg
```

```python
genre_weekday(moscow_general, 'Friday', '17:00', '23:00') # call the function for Friday evening in Moscow 
```

```python
genre_weekday(spb_general, 'Friday', '17:00', '23:00') # call the function for Friday evening in St. Petersburg
```

**Conclusions**

If we compare the top 10 genres on Monday morning, we can draw the following conclusions:

1. In Moscow and St. Petersburg, people listen to similar music. The only difference is that the Moscow ranking includes the "world" genre, while the St. Petersburg ranking includes jazz and classical music.

2. In Moscow, there were so many missing values that the value `'unknown'` ranked tenth among the most popular genres. This means that missing values occupy a significant share of the data and threaten the validity of the study.

Friday evening does not change this picture. Some genres move slightly up, others move down, but overall the top 10 remains the same.

Thus, the second hypothesis was only partially confirmed:
* Users listen to similar music at the beginning and end of the week.
* The difference between Moscow and St. Petersburg is not very pronounced. In Moscow, Russian pop music is listened to more frequently, while in St. Petersburg, jazz is more popular.

However, the missing values cast doubt on this result. In Moscow, there are so many missing values that the top 10 ranking could look different if the lost data on genres were available.
